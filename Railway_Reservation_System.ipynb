```python
#lab16.
#1.
import re

def validate_pnr(pnr):
    pattern = r'^\d{8}$'
    if re.fullmatch(pattern, pnr):
        print("Valid PNR :", pnr)
    else:
        print("Invalid PNR :", pnr)
```

```python
#2.
validate_pnr("43821756")   # Valid
validate_pnr("6793452")    # Invalid - only 7 digits
validate_pnr("8120AB63")   # Invalid - contains letters
validate_pnr("81204763")   # Valid
```

```python
#3.
import re

def validate_seat(seat):
    pattern = r'^[A-Z][0-9]{1,2}$'
    if re.fullmatch(pattern, seat):
        print("Valid Seat Number :", seat)
    else:
        print("Invalid Seat Number :", seat)
```

```python
#4.
validate_seat("B4")    # Valid
validate_seat("A12")   # Valid
validate_seat("D23")   # Valid
validate_seat("b4")    # Invalid - lowercase letter
validate_seat("23B")   # Invalid - wrong order
validate_seat("A")     # Invalid - missing digit
```

```python
#5.
import re

def extract_train_numbers(text):
    pattern = r'\b[0-9]{4,5}\b'
    results = re.findall(pattern, text)
    print("Train Numbers Found :", results)
    return results
```

```python
#7.
import re

class TicketValidator:

    def __init__(self, pnr, train_no, seat, fare):
        self.pnr      = pnr
        self.train_no = train_no
        self.seat     = seat
        self.fare     = fare

    def validate(self):
        print("=" * 40)
        print("   IndRail Ticket Validation Report")
        print("=" * 40)

        # Validate PNR - exactly 8 digits
        if re.fullmatch(r'[0-9]{8}', self.pnr):
            print("PNR      :", self.pnr, "-> VALID")
        else:
            print("PNR      :", self.pnr, "-> INVALID")

        # Validate Train Number - 4 or 5 digits
        if re.fullmatch(r'[0-9]{4,5}', self.train_no):
            print("Train No :", self.train_no, "-> VALID")
        else:
            print("Train No :", self.train_no, "-> INVALID")

     
```

```python
#8.
ticket1 = TicketValidator("43821756", "12951", "B4",  "880")
ticket2 = TicketValidator("6793452",  "1230X", "b12", "XX")

ticket1.validate()
ticket2.validate()
```

```python
#3.
import requests
from bs4 import BeautifulSoup

url = "https://contentitvedant.github.io/Railway_Booking/trains.html"

response = requests.get(url)
soup     = BeautifulSoup(response.text, "html.parser")

table = soup.find(id="train-schedule-table")
rows  = table.find("tbody").find_all("tr")

print("Train Schedule:")
print("=" * 60)

for row in rows:
    cells      = row.find_all("td")
    train_no   = cells[0].get_text()
    train_name = cells[1].get_text()
    from_stn   = cells[2].get_text()
    to_stn     = cells[3].get_text()
    departure  = cells[4].get_text()
    arrival    = cells[5].get_text()
    duration   = cells[6].get_text()

    print(train_no, "|", train_name)
    print("   From :", from_stn, "-> To :", to_stn)
    print("   Departure :", departure, "| Arrival :", arrival, "| Duration :", duration)
    print("-" * 60)
```

```python
#6.
import requests
from bs4 import BeautifulSoup

class RailwayScraper:

    def __init__(self):
        self.base_url = "https://contentitvedant.github.io/Railway_Booking/"

    def get_soup(self, page):
        url      = self.base_url + page
        response = requests.get(url)
        soup     = BeautifulSoup(response.text, "html.parser")
        print("Fetched :", url, "| Status :", response.status_code)
        return soup

    def get_train_schedule(self):
        soup   = self.get_soup("trains.html")
        table  = soup.find(id="train-schedule-table")
        rows   = table.find("tbody").find_all("tr")
        trains = []
        for row in rows:
            cells = row.find_all("td")
            trains.append({
                "train_no"   : cells[0].get_text(),
                "train_name" : cells[1].get_text(),
                "from"       : cells[2].get_text(),
                "to"         : cells[3].get_text(),
                "departure"  : cells[4].get_text(),
                "arrival"    : cells[5].get_text(),
                "duration"   : cells[6].get_text(),
            })
        return trains

    def get_passengers(self):
        soup       = self.get_soup("passengers.html")
        table      = soup.find(id="passenger-records-table")
        rows       = table.find("tbody").find_all("tr")
        passengers = []
        for row in rows:
            passengers.append({
                "pnr"  : row["data-pnr"],
                "name" : row.find(class_="passenger-name").get_text(),
                "date" : row.find(class_="journey-date").get_text(),
                "class": row.find(class_="travel-class").get_text(),
                "seat" : row.find(class_="seat-number").get_text(),
                "fare" : row.find(class_="fare").get_text(),
            })
        return passengers

    def get_seat_availability(self):
        soup      = self.get_soup("seat-availability.html")
        table     = soup.find(id="seat-availability-table")
        rows      = table.find("tbody").find_all("tr")
        available = []
        for row in rows:
            available.append({
                "train_no"   : row["data-train"],
                "train_name" : row.find(class_="train-name").get_text(),
                "seats_sl"   : row.find(class_="seats-sl").get_text(),
                "seats_3a"   : row.find(class_="seats-3a").get_text(),
                "seats_2a"   : row.find(class_="seats-2a").get_text(),
                "seats_1a"   : row.find(class_="seats-1a").get_text(),
            })
        return available
```

```python
#7.
scraper = RailwayScraper()

print("\n--- TRAIN SCHEDULE ---")
trains = scraper.get_train_schedule()
for t in trains:
    print(t["train_no"], "|", t["train_name"], "|", t["from"], "->", t["to"], "|", t["duration"])

print("\n--- PASSENGER RECORDS ---")
passengers = scraper.get_passengers()
for p in passengers:
    print("PNR:", p["pnr"], "| Name:", p["name"], "| Class:", p["class"], "| Seat:", p["seat"])

print("\n--- SEAT AVAILABILITY ---")
seats = scraper.get_seat_availability()
for s in seats:
    print("Train", s["train_no"], "-", s["train_name"], "| SL:", s["seats_sl"], "| 3A:", s["seats_3a"], "| 2A:", s["seats_2a"])
```

```python
#8.
file = open("train_schedule.txt", "w")

for t in trains:
    file.write(
        t["train_no"] +
        " - " +
        t["train_name"] +
        "\n"
    )

file.close()

print("Train schedule saved successfully.")
```

```python
#lab18.
#1.
import re

class RailwayReservationSystem:

    def __init__(self):
        self.train_name = "Rajdhani Express"
        self.available_seats = 5
        self.passengers = []

    def book_ticket(self, name, pnr, seat, fare):

        passenger = {
            "name": name,
            "pnr": pnr,
            "seat": seat,
            "fare": fare
        }

        self.passengers.append(passenger)

    def validate_ticket(self, pnr, seat):

        pnr_pattern = r'^\d{8}$'
        seat_pattern = r'^[A-Z][0-9]{1,2}$'

        pnr_valid = re.fullmatch(pnr_pattern, pnr)
        seat_valid = re.fullmatch(seat_pattern, seat)

        return pnr_valid and seat_valid
 
    def allocate_seat(self):
 
        if self.available_seats > 0:
            self.available_seats -= 1
            return "Seat Allocated"
 
        return "No Seats Available"
```

```python
#2.
system = RailwayReservationSystem()
```

```python
#3.
system.book_ticket(
   "Rahul Sharma",
   "43821756",
   "A1",
   1200
)

system.book_ticket(
   "Priya Singh",
   "67934521",
   "A2",
   1200
)

print(system.passengers)
```

```python
#4.
for passenger in system.passengers:

   result = system.validate_ticket(
       passenger["pnr"],
       passenger["seat"])

   print(
       passenger["name"],
       "-",
       result)
```

```python
#5.
for passenger in system.passengers:

   print(
       passenger["name"],
       "-",
       system.allocate_seat())
```

```python
#6.
print(
   "Available Seats :",
   system.available_seats)
```

```python
#7.
for passenger in system.passengers:

    print("=" * 40)

    print("Passenger :", passenger["name"])
    print("PNR       :", passenger["pnr"])
    print("Seat      :", passenger["seat"])
    print("Fare      :", passenger["fare"])

    print("=" * 40)

print("\nBooking Summary")
print("-" * 30)

print(
    "Total Passengers :",
    len(system.passengers))

print(
    "Available Seats :",
    system.available_seats)

if system.available_seats > 0:
    print("Reservation Status : Open")
else:
    print("Reservation Status : Full")
```
